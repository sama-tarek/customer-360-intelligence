# Customer 360 - Data Overview

## Business Problem

The company wants to better understand its customers and make better decisions using customer data.

The main goals are to:

* Predict which customers are likely to **churn**.
* Predict each customer's **future 12-month revenue**.
* Group customers into different **customer segments**.
* Detect **unusual customer behavior**.
* Combine these results into one **Customer 360 view**.

This can help the company understand customer behavior, identify customers who may need attention, and better understand customer value.


## Project Objectives

In this project, we will:

* Explore and clean the customer data.
* Analyze customer behavior and identify important patterns.
* Build models to predict churn and future revenue.
* Segment customers using clustering.
* Detect unusual customers using anomaly detection.
* Use PCA to visualize customer patterns.
* Combine the ML results into a final Customer 360 table.

## Dataset Column Description

The dataset contains customer demographic, subscription, usage, service, payment, satisfaction, churn, and future revenue information. Each row represents one customer.

| Column                  | Description                                                                                                                                                                                                          |
| ----------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| `CustomerID`          | Unique ID for each customer.|
| `Age`                 | Customer's age.                                                                                |
| `Region`              | Customer's region.|
| `TenureMonths`        | Number of months the customer has been with the company.                                                                           |
| `ContractType`        | Customer's contract type.                                                 |
| `InternetType`        | Type of internet service used by the customer.                                                                |
| `MonthlyUsageGB`      | Customer's monthly internet usage in GB.                                                                                          |
| `MonthlyChargeEGP`    | Customer's monthly charge in Egyptian Pounds.                                                                      |
| `NumServices`         | Number of services used by the customer.                                                                      |
| `SupportCalls6M`      | Number of support calls in the last 6 months.                                                          |
| `LatePayments12M`     | Number of late payments in the last 12 months.                                                                                   |
| `SatisfactionScore`   | Customer satisfaction score.                                                                                    |
| `AutoPay`             | Whether the customer uses automatic payment.                                                                                |
| `Churn`               | Whether the customer left the service. Used as the classification target.                                                           |
| `Future12MRevenueEGP` | Customer's revenue over the next 12 months. Used as the regression target. |

### Target Variables

This project contains two supervised learning tasks:

* **Churn Classification:** Predict whether a customer will churn using `Churn` as the target.
* **Revenue Regression:** Predict the customer's future 12-month revenue using `Future12MRevenueEGP` as the target.

### Identifier Column

`CustomerID` is retained for customer-level reporting and for constructing the final **Customer 360** table, but it is excluded from machine learning models because it is an identifier rather than a meaningful predictive feature.
