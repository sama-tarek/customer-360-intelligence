# Customer 360 - Data Cleaning

This notebook focuses on cleaning and preparing the collected customer data before performing Exploratory Data Analysis (EDA).

### Objectives
- Identify missing values
- Check duplicated records
- Handle inconsistent values
- Fix data types
- Handle missing values
- Check for invalid or unrealistic values
- Save the cleaned dataset for the next stage

## Import Libararies

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

## Load Dataset

```python
df = pd.read_csv('../data/raw/customer_360_ml_workshop.csv')
```

## Data Cleaning

### Check duplicates

```python
df.duplicated().sum()
```

The dataset doesn't have **duplicates**.

### Check missing values

```python
missing_count = (df.isnull().sum()).sort_values(ascending= False)

missing_percentage = (df.isnull().mean() * 100).sort_values(ascending= False)

missing_summary = pd.DataFrame({
    'Missing Count' : missing_count[missing_count > 0] , 
    'Missing Percentage' : missing_percentage[missing_percentage > 0]
})

missing_summary
```

The dataset contains a small number of missing values:

* `MonthlyUsageGB`: 75 missing values (2.5%)
* `SatisfactionScore`: 60 missing values (2.0%)
* `InternetType`: 45 missing values (1.5%)

Since the percentage of missing values is low, the affected rows will not be removed.

* Numerical columns(`MonthlyUsageGB` , `SatisfactionScore`) will be filled using the **median**.
* The categorical column(`InternetType`) will be filled using the **most frequent value (mode)**.


### Missing Value Treatment

```python
df['MonthlyUsageGB'] = df['MonthlyUsageGB'].fillna(df['MonthlyUsageGB'].median())
df['SatisfactionScore'] = df['SatisfactionScore'].fillna(df['SatisfactionScore'].median())

df['InternetType'] = df['InternetType'].fillna(df['InternetType'].mode()[0])
```

### Check impossible values

```python
print(f'Impossible Age Values: {len(df[df["Age"] < 0])}')
print(f'Impossible TenureMonths Values: {len(df[df["TenureMonths"] < 0])}')
print(f'Impossible MonthlyChargeEGP Values: {len(df[df["MonthlyChargeEGP"] < 0])}')
print(f'Impossible MonthlyUsageGB Values: {len(df[df["MonthlyUsageGB"] < 0])}')
print(f'Impossible NumServices Values: {len(df[df["NumServices"] < 0])}')
print(f'Impossible SupportCalls6M Values: {len(df[df["SupportCalls6M"] < 0])}')
print(f'Impossible LatePayments12M Values: {len(df[df["LatePayments12M"] < 0])}')
```

The dataset doesn't have **Impossible Values**.

### Final Missing Value Check

After handling missing and invalid values, we perform a final check to make sure the dataset is ready for EDA.

```python
missing_summary_final = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_summary_final[
    missing_summary_final["Missing Values"] > 0
]
```
